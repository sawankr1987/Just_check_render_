```python
# ----------------------------
# Imports and Compatibility Fixes
# ----------------------------
import numpy as np

# Fix NumPy 2.0 compatibility
if not hasattr(np, 'bool'):
    np.bool = bool          # built-in bool
if not hasattr(np, 'complex_'):
    np.complex_ = np.complex128
if not hasattr(np, 'float_'):
    np.float_ = np.float64
if not hasattr(np, 'string_'):
    np.string_ = np.bytes_
if not hasattr(np, 'unicode_'):
    np.unicode_ = np.str_


import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
import joblib

# ----------------------------
# Load dataset
# ----------------------------
df = pd.read_csv('diabetes.csv')  # Ensure this file exists in your working directory

# ----------------------------
# Split features and target
# ----------------------------
X = df.drop('Outcome', axis=1)
y = df['Outcome']

# ----------------------------
# Scale features
# ----------------------------
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# ----------------------------
# Train-test split
# ----------------------------
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42
)


```

```python
# ----------------------------
# Build model
# ----------------------------
model = Sequential([
    Dense(16, input_dim=X_train.shape[1], activation='relu'),
    Dense(8, activation='relu'),
    Dense(1, activation='sigmoid')
])

# ----------------------------
# Compile and train
# ----------------------------
model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
model.fit(X_train, y_train, epochs=100, batch_size=10, verbose=0)

# ----------------------------
# Save model and scaler
# ----------------------------
model.save('model_tf.h5')
joblib.dump(scaler, 'scaler.save')

print("✅ TensorFlow model and scaler saved successfully as 'model_tf.h5' and 'scaler.save'.")
```

```python
from fmpy import *
import numpy as np

class ScalerFMU:
    def __init__(self):
        # Extracted scaler parameters
        self.mean = np.array([3.8, 120.9, 69.1, 20.5, 79.8, 31.9, 0.47, 33.2])
        self.scale = np.array([3.4, 31.0, 19.4, 15.9, 115.2, 7.9, 0.33, 11.8])
        self.features = np.zeros(8)          # dtype left to default (float)
        self.scaled_features = np.zeros(8)  # dtype float

    def initialize(self):
        pass

    def do_step(self, current_time, step_size):
        self.scaled_features = (self.features - self.mean) / self.scale

    def set_input(self, input_values):
        self.features = np.array(input_values)

    def get_output(self):
        return self.scaled_features.tolist()
```
